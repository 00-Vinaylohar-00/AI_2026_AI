```python
l=['kedar','mandar','anup','rushi','shital']
l
```

```python
v=(i.upper() for i in l)
list(v)
```

```python
v=map(lambda i:i.upper(),l)
list(v)
```

```python
num =(1,5,9,7,-3,-5,-7,-3)
v=map(lambda i:f"{i} positive" if i>0 else f"{i} negative",num)
",".join(list(v))
```

```python
lst=[1,2,3,4,5,8,7]
list(map(lambda x:x**2,lst))
```

```python
lst=["India is my country"]

list(map(lambda x:x.split(),lst))
```

```python
lst=[1,2,3,4,5,8,7]
list(filter(lambda x:x%2!=0,lst))
```

```python
data = [1, "apple", 3.14, "banana", True, 42]
x=list(filter(lambda x: isinstance(x, str), data))
print(x)
list(filter(lambda x: isinstance(x, (int, float)), data))
list(filter(lambda x: isinstance(x,complex), data))
# isinstance() working on dtatypes if given dta type is present in list the it will give that data as result
# if no data is their then it will provides empty
```
