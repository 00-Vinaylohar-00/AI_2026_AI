## **To work on CO2 emission dataset**

```python
path = 'https://raw.githubusercontent.com/hakimyameen/Datasets/main/CO2%20Emissions_Canada.csv'
path
```

```python
# import ds libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# read the dataset
file = pd.read_csv(path)
file[:2]
```

```python
# check number of rows n columns
file.shape
```

```python
# display column names
#file.info()
file.columns
```

```python
# summary of dataset
# Print a concise summary of a DataFrame.
file.info()
```

```python
#from typing_extensions import Type
# select any one column
file['Cylinders']
file['Make']
file['Fuel Type']
```

```python
# select 2 or more columns
file[['CO2 Emissions(g/km)','Make','Transmission']]
```

```python
# Check distinct records of Model column
file['Model'].drop_duplicates() # distnct model means without duplicates
```

```python
# count unique models
file['Model'].nunique() # unique gives the array data regardslesss of duplicates
```

```python
# display per model count
file["Model"].value_counts()
```

```python
# display top 10 models with its count
y=file['Model'].value_counts().head(10)
y
```

```python
# plot per model count
#y.plot(kind='bar')
x=file['Model'].value_counts().head(10).index # used to show limited data with index must be their
sns.countplot(x='Model',data=file,color='skyblue',order=x) # order is used to display only limited and required data
plt.xticks(rotation=45)
plt.figure(figsize=(12,6))
plt.show()
```

```python
# sort Model column A-Z
file['Model'].sort_values(ascending=True) # sorting only model coloumn
file.sort_values(by='Model',ascending=True) # sorting whole data frame by model in acending
```

```python
# select model names starting with M
file[file['Model'].str.startswith('M')]['Model']
```

```python
# check data types of columns
file.columns.inferred_type # gives single output as str or mixed data or int or bool data
file.dtypes  # Gives whole framework dta where it is object, bool , float, int
```

```python
# select object columns
file.select_dtypes(include='object')
```

```python
# Select the records where 'Vehicle Class' is SUV - SMALL
file[file['Vehicle Class'].str.startswith('SUV - SMALL')][['Model','Vehicle Class']]
#file[file['Vehicle Class'] == 'SUV - SMALL']
```

```python
# display count of unique records present in each object column
file.select_dtypes(include=object).nunique()
```

```python
# display the records of a vehicle whose CO2 Emissions(g/km) is max
file.loc[file['CO2 Emissions(g/km)'].idxmax(), ['CO2 Emissions(g/km)', 'Model']] # to get indivisual max data with additional 2D table given
file.loc[file['CO2 Emissions(g/km)'].idxmax()] # to get max full frame Data idmax() is used.
file.sort_values(by='CO2 Emissions(g/km)', ascending=False).head()[['Make','Model','Vehicle Class','CO2 Emissions(g/km)']]
# above query is with using sort method
file[file['CO2 Emissions(g/km)'] == file['CO2 Emissions(g/km)'].max()][['Make','Model','Vehicle Class','CO2 Emissions(g/km)']]
# above query displays multiple records in frame
file.iloc[file['CO2 Emissions(g/km)'].values.argmax()][['Make','Model','Vehicle Class','CO2 Emissions(g/km)']]
# with iloc method
```

```python
# display the records of a vehicle whose CO2 Emissions(g/km) is greater than average CO2 Emissions(g/km)
file[file['CO2 Emissions(g/km)']>file['CO2 Emissions(g/km)'].mean()]
```
