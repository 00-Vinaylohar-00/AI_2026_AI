```python

k = [9,8,7,6,5,4]

for i in range(len(k)):
    print(" " * (len(k)-i-1) + "*" * (2*i+1))
        # spaces * (6-0-1) * + * into (2*0+1)
        # 5 spaces both side   +  1 star
        # loop iterate upto lenth
```

```python
k = [9,8,7,6,5,4]
n = len(k)

for i in range(n):
    spaces = n - i - 1
    stars = 2*i + 1

    print(f"{n}-{i}-1 = {spaces} spaces")
    print(f"2*{i}+1 = {stars} stars")
    print(" " * spaces + "*" * stars)
    print()
```

```python
data = [
    [1, 2, 3],
    ("a", "b", "c"),
    "xyz",
    (10, 20)
]

for i in data:
    if isinstance(i, (list, tuple, str)): #checking the data tyeps
        for j in i: # going inside i printing every thing in i as like 1,2,3
            print(j, end=" ")
        print()
    else: # it won't run as our loop iterates in data correctly
        print(i)
```

```python
y={"virat",45,"sharad",4+5j,True,3.14,67}
a=[]
b=[]
c=[]
d=[]
for items in y:
  if isinstance(items,str):
    a.append(items)
  elif isinstance(items,int):
    b.append(items)
  elif isinstance(items,complex):
    c.append(items)
  else:
    d.append(items)
print(a)
print(b)
print(c)
print(d)
```

```python
z={'name':"Vinay",'age':26,'float':3.15,'value':True,"comp":5+8j}
for key, value in z.items():
  if isinstance(value,int):
    print(key,"integers:",value)
  elif isinstance(value,str):
    print(key,"string:",value)
  elif isinstance(value,complex):
    print(key,"complex:",value)
  elif isinstance(value,bool):
    print(key,"boolean:",value)
  else:
    print(key,"Float:",value)
```
