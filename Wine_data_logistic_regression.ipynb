```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_wine
```

```python
w=load_wine()
```

```python
w.keys()
```

```python
w['target_names']
```

```python
load_wine()
```

```python
w['feature_names']
```

```python
x=pd.DataFrame(w['data'],columns=w['feature_names'])
x[:2]
```

```python
y=w['target']
y[:2]
```

```python
x.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(x,y,train_size=0.7)
```

```python
X_train[:5]
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
Y_train[:5]
```

```python
model=LogisticRegression()
model
```

```python
model.fit(X_train,Y_train)
```

```python
X_test[:3]
```

```python
Y_test[:20]
```

```python
y_pred=model.predict(X_test)
y_pred
```

```python
model.score(X_train,Y_train)*100
```

```python
from sklearn.metrics import confusion_matrix
```

```python
cm=confusion_matrix(Y_test,y_pred)
cm
```

```python
plt.figure(figsize=(5,5))
sns.heatmap(cm,annot=True, xticklabels=['class_0','class_1','class_2'],yticklabels=['class_0','class_1','class_2'])
```

```python
# Class 0
print(sum((Y_test==0) & (y_pred==0)))  # True class 0 predicted as 0 TP
print(sum((Y_test==0) & (y_pred==1)))  # Class 0 misclassified as 1  FN
print(sum((Y_test==0) & (y_pred==2)))  # Class 0 misclassified as 2  FN

# Class 1
print(sum((Y_test==1) & (y_pred==0)))  # Class 1 misclassified as 0  FP
print(sum((Y_test==1) & (y_pred==1)))  # True class 1 predicted as 1 TN
print(sum((Y_test==1) & (y_pred==2)))  # Class 1 misclassified as 2  TN negative predicted also negative

# Class 2
print(sum((Y_test==2) & (y_pred==0)))  # Class 2 misclassified as 0  FP
print(sum((Y_test==2) & (y_pred==1)))  # Class 2 misclassified as 1  TN negative predicted as negative
print(sum((Y_test==2) & (y_pred==2)))  # True class 2 predicted as 2 TN
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
```

```python
accuracy_score(Y_test,y_pred)*100
```

```python
precision_score(Y_test,y_pred, average='weighted')*100
```

```python
recall_score(Y_test,y_pred, average='weighted')*100
```

```python
f1_score(Y_test,y_pred, average='weighted')*100
```

```python
from sklearn.metrics import classification_report
print(classification_report(Y_test,y_pred))
```
